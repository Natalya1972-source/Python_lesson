ДЗ 10.2. Знайти перше слово


1. Реалізовано функцію для пошуку першого слова в рядку з урахуванням розділових знаків та апострофа як частини слова.

```python
def first_word(text):
    """ Пошук першого слова """
    word = ""
    for char in text:
        if char.isalpha() or char == "'":
            word += char
        elif word:
            break
    return word


# Тести
assert first_word("Hello world") == "Hello", 'Test1'
assert first_word("greetings, friends") == "greetings", 'Test2'
assert first_word("don't touch it") == "don't", 'Test3'
assert first_word(".., and so on ...") == "and", 'Test4'
assert first_word("hi") == "hi", 'Test5'
assert first_word("Hello.World") == "Hello", 'Test6'
print('OK')
```

2. через split та replace

```python
def first_word(text):
    """ Пошук першого слова """
    text = text.replace('.', ' ').replace(',', ' ')
    return text.split()[0]


# Перевірки
assert first_word("Hello world") == "Hello"
assert first_word("greetings, friends") == "greetings"
assert first_word("don't touch it") == "don't"
assert first_word(".., and so on ...") == "and"
assert first_word("hi") == "hi"
assert first_word("Hello.World") == "Hello"
print("OK")
```

3. через генератор

```python
def first_word(text):
    """ Пошук першого слова """
    cleaned = (c if c.isalpha() or c == "'" else " " for c in text)
    return "".join(cleaned).split()[0]


# Перевірки
assert first_word("Hello world") == "Hello"
assert first_word("greetings, friends") == "greetings"
assert first_word("don't touch it") == "don't"
assert first_word(".., and so on ...") == "and"
assert first_word("hi") == "hi"
assert first_word("Hello.World") == "Hello"
print("OK")
```
