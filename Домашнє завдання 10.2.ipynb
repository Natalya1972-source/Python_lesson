ДЗ 10.1. Генераторна функція


1. Реалізовано генераторну функцію, яка за заданим початковим значенням, кількістю елементів і користувацькою функцією поетапно формує числову послідовність.

```python
def pow(x):
    return x ** 2

def some_gen(begin, end, func):
    """
     begin: перший елемент послідовності
     end: кількість елементів у послідовності
     func: функція, яка формує значення для послідовності
    """
    current = begin
    for _ in range(end):
        yield current
        current = func(current)

from inspect import isgenerator

gen = some_gen(2, 4, pow)
assert isgenerator(gen) == True, 'Test1'
assert list(gen) == [2, 4, 16, 256], 'Test2'
print('OK')
```

2. через while

```python
def pow(x):
    return x ** 2

def some_gen(begin, end, func):
    """
     begin: перший елемент послідовності
     end: кількість елементів у послідовності
     func: функція, яка формує значення для послідовності
    """
    current = begin
    count = 0
    while count < end:
        yield current
        current = func(current)
        count += 1

from inspect import isgenerator

gen = some_gen(2, 4, pow)
assert isgenerator(gen) == True, 'Test1'
assert list(gen) == [2, 4, 16, 256], 'Test2'
print('OK')
```
