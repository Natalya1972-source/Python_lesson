ДЗ 11.2. Заповнення списку кубами чисел


1. Варіант. Функція-генератор по черзі обчислює куби чисел, починаючи з 2, і повертає їх, поки значення не перевищує задану межу

```python
def generate_cube_numbers(end):
    num = 2
    while True:
        cube = num ** 3
        if cube > end:
            return
        yield cube
        num += 1

from inspect import isgenerator

gen = generate_cube_numbers(1)
assert isgenerator(gen) == True, 'Test0'
assert list(generate_cube_numbers(10)) == [8]
assert list(generate_cube_numbers(100)) == [8, 27, 64]
assert list(generate_cube_numbers(1000)) == [8, 27, 64, 125, 216, 343, 512, 729, 1000]
print('Ok')
```

2. Варіант. Генератор перебирає числа від 2, обчислює їх куби та повертає значення, поки вони не перевищують задану межу.

```python
def generate_cube_numbers(end):
    for num in range(2, end):
        cube = num ** 3
        if cube > end:
            return
        yield cube

from inspect import isgenerator

gen = generate_cube_numbers(1)
assert isgenerator(gen) == True, 'Test0'
assert list(generate_cube_numbers(10)) == [8]
assert list(generate_cube_numbers(100)) == [8, 27, 64]
assert list(generate_cube_numbers(1000)) == [8, 27, 64, 125, 216, 343, 512, 729, 1000]
print('Ok')
```
