ДЗ 11.3. Перевірка на парність.


1. Функція перевіряє парність числа за допомогою побітової операції, визначаючи значення останнього біта без використання ділення.

```python
def is_even(number):
    return (number & 1) == 0

assert is_even(2494563894038**2) == True, 'Test1'
assert is_even(1056897**2) == False, 'Test2'
assert is_even(24945638940387**3) == False, 'Test3'
print("Ok")
```

2. Через двійковий рядок. Останній символ двійкового представлення показує парність. Парні числа в двійковій системі закінчуються на 0.

```python
def is_even(number):
    return bin(number)[-1] == '0'

assert is_even(2494563894038**2) == True, 'Test1'
assert is_even(1056897**2) == False, 'Test2'
assert is_even(24945638940387**3) == False, 'Test3'
print("Ok")
```

3. Через XOR (або , но не разом). 

```python
def is_even(number):
    return (number ^ 1) == number + 1

assert is_even(2494563894038**2) == True, 'Test1'
assert is_even(1056897**2) == False, 'Test2'
assert is_even(24945638940387**3) == False, 'Test3'
print("Ok")
```
