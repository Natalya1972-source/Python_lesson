ДЗ 12.1. Очистити текст від html-тегів

1. Варіант. Ця функція читає HTML-файл, посимвольно видаляє всі теги разом із вмістом між «<» і «>», прибирає порожні рядки та записує очищений текст у новий файл.

```python
import codecs


def delete_html_tags(html_file, result_file='cleaned.txt'):
    with codecs.open(html_file, 'r', 'utf-8') as file:
        html = file.read()

    cleaned_text = ''
    inside_tag = False

    for char in html:
        if char == '<':
            inside_tag = True
        elif char == '>':
            inside_tag = False
        elif not inside_tag:
            cleaned_text += char

    # Прибираємо порожні рядки
    lines = cleaned_text.splitlines()
    cleaned_lines = [line.strip() for line in lines if line.strip()]

    with codecs.open(result_file, 'w', 'utf-8') as file:
        file.write('\n'.join(cleaned_lines))
```

```python
delete_html_tags('draft.html')
```

2. Варіант (через регулярні вирази). Ця функція читає HTML-файл, видаляє всі теги за допомогою регулярного виразу, очищає текст від порожніх рядків і записує результат у новий файл.

```python
import re


def delete_html_tags(html_file, result_file='cleaned.txt'):
    with codecs.open(html_file, 'r', 'utf-8') as file:
        html = file.read()

    # Видаляємо всі HTML-теги
    cleaned_text = re.sub(r'<[^>]+>', '', html)

    # Прибираємо зайві пробіли та порожні рядки
    lines = cleaned_text.splitlines()
    cleaned_lines = [line.strip() for line in lines if line.strip()]

    with codecs.open(result_file, 'w', 'utf-8') as file:
        file.write('\n'.join(cleaned_lines))
```

3. Варіант- Цей варіант використовує клас HTMLParser для коректного вилучення тексту з HTML-документа, після чого очищає його від порожніх рядків і записує результат у новий файл.

```python
import codecs
from html.parser import HTMLParser


class HTMLCleaner(HTMLParser):
    def __init__(self):
        super().__init__()
        self.text = []

    def handle_data(self, data):
        self.text.append(data)


def delete_html_tags(html_file, result_file='cleaned.txt'):
    with codecs.open(html_file, 'r', 'utf-8') as file:
        html = file.read()

    parser = HTMLCleaner()
    parser.feed(html)

    cleaned_text = ''.join(parser.text)

    # Прибираємо порожні рядки
    lines = cleaned_text.splitlines()
    cleaned_lines = [line.strip() for line in lines if line.strip()]

    with codecs.open(result_file, 'w', 'utf-8') as file:
        file.write('\n'.join(cleaned_lines))
```

```python
delete_html_tags('draft.html')
```
