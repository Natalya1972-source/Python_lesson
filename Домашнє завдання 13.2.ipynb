ДЗ 13.2. Клас "Цифровий лічильник"

1. Цей код реалізує клас цифрового лічильника з мінімальним, максимальним і поточним значенням, який дозволяє змінювати межі, коректно збільшувати або зменшувати значення в допустимому діапазоні та викидає виняток ValueError при досягненні встановлених меж.

```python
class Counter:

    def __init__(self, current=1, min_value=0, max_value=10):
        if min_value > max_value:
            raise ValueError("Мінімум не може бути більшим за максимум")
        if not (min_value <= current <= max_value):
            raise ValueError("Поточне значення має бути в межах мінімуму та максимуму")

        self.current = current
        self.min_value = min_value
        self.max_value = max_value

    def set_current(self, start):
        if not (self.min_value <= start <= self.max_value):
            raise ValueError("Початкове значення має бути в межах мінімуму та максимуму")
        self.current = start

    def set_max(self, max_max):
        if max_max < self.min_value:
            raise ValueError("Максимум не може бути меншим за мінімум")
        self.max_value = max_max
        if self.current > self.max_value:
            self.current = self.max_value  # щоб лічильник не виходив за межі

    def set_min(self, min_min):
        if min_min > self.max_value:
            raise ValueError("Мінімум не може бути більшим за максимум")
        self.min_value = min_min
        if self.current < self.min_value:
            self.current = self.min_value  # щоб лічильник не виходив за межі

    def step_up(self):
        if self.current >= self.max_value:
            raise ValueError("Досягнуто максимуму")
        self.current += 1

    def step_down(self):
        if self.current <= self.min_value:
            raise ValueError("Досягнутий мінімум")
        self.current -= 1

    def get_current(self):
        return self.current


counter = Counter()
counter.set_current(7)
counter.step_up()
counter.step_up()
counter.step_up()
assert counter.get_current() == 10, 'Test1'
try:
    counter.step_up()  # ValueError
except ValueError as e:
    print(e)  # Досягнуто максимуму
assert counter.get_current() == 10, 'Test2'

counter.set_min(7)
counter.step_down()
counter.step_down()
counter.step_down()
assert counter.get_current() == 7, 'Test3'
try:
    counter.step_down()  # ValueError
except ValueError as e:
    print(e)  # Досягнутий мінімум
assert counter.get_current() == 7, 'Test4'
```

2. Цей варіант реалізує цифровий лічильник із централізованою перевіркою коректності діапазону значень, забезпечуючи контроль меж та генерацію винятків при досягненні мінімального або максимального значення.

```python
class Counter:

    def __init__(self, current=1, min_value=0, max_value=10):
        self.min_value = min_value
        self.max_value = max_value
        self._validate_range()
        self.set_current(current)

    def _validate_range(self):
        if self.min_value > self.max_value:
            raise ValueError("Мінімум не може бути більшим за максимум")

    def set_current(self, value):
        if not self.min_value <= value <= self.max_value:
            raise ValueError("Поточне значення поза межами допустимого діапазону")
        self.current = value

    def set_max(self, value):
        self.max_value = value
        self._validate_range()
        if self.current > self.max_value:
            self.current = self.max_value

    def set_min(self, value):
        self.min_value = value
        self._validate_range()
        if self.current < self.min_value:
            self.current = self.min_value

    def step_up(self):
        if self.current == self.max_value:
            raise ValueError("Досягнуто максимуму")
        self.current += 1

    def step_down(self):
        if self.current == self.min_value:
            raise ValueError("Досягнутий мінімум")
        self.current -= 1

    def get_current(self):
        return self.current
    
counter = Counter()
counter.set_current(7)
counter.step_up()
counter.step_up()
counter.step_up()
assert counter.get_current() == 10, 'Test1'
try:
    counter.step_up()  # ValueError
except ValueError as e:
    print(e)  # Досягнуто максимуму
assert counter.get_current() == 10, 'Test2'

counter.set_min(7)
counter.step_down()
counter.step_down()
counter.step_down()
assert counter.get_current() == 7, 'Test3'
try:
    counter.step_down()  # ValueError
except ValueError as e:
    print(e)  # Досягнутий мінімум
assert counter.get_current() == 7, 'Test4'


    
```
