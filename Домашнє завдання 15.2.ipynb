1. Я реалізувала клас дробів і перевизначила магічні методи арифметичних операцій та порівняння, використовуючи стандартні формули роботи з дробами.

```python
class Fraction:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def __mul__(self, other):
        return Fraction(self.a * other.a, self.b * other.b)

    def __add__(self, other):
        new_a = self.a * other.b + other.a * self.b
        new_b = self.b * other.b
        return Fraction(new_a, new_b)

    def __sub__(self, other):
        new_a = self.a * other.b - other.a * self.b
        new_b = self.b * other.b
        return Fraction(new_a, new_b)

    def __eq__(self, other):
        return self.a * other.b == other.a * self.b

    def __gt__(self, other):
        return self.a * other.b > other.a * self.b

    def __lt__(self, other):
        return self.a * other.b < other.a * self.b

    def __str__(self):
        return f"Fraction: {self.a}, {self.b}"


# тест
f_a = Fraction(2, 3)
f_b = Fraction(3, 6)

f_c = f_b + f_a
assert str(f_c) == 'Fraction: 21, 18'

f_d = f_b * f_a
assert str(f_d) == 'Fraction: 6, 18'

f_e = f_a - f_b
assert str(f_e) == 'Fraction: 3, 18'

assert f_d < f_c
assert f_d > f_e
assert f_a != f_b

f_1 = Fraction(2, 4)
f_2 = Fraction(3, 6)
assert f_1 == f_2

print("OK")
```

2. Код створює клас Fraction для роботи з дробами, який автоматично скорочує дріб через НСД (math.gcd) під час створення та перевизначає магічні методи для додавання, віднімання, множення і порівняння дробів, повертаючи нові скорочені дроби.

```python
import math

class Fraction:
    def __init__(self, a, b):
        if b == 0:
            raise ValueError("Знаменник не може бути 0")

        gcd = math.gcd(a, b)
        self.a = a // gcd
        self.b = b // gcd
        print(f"Створено дріб: {self.a}/{self.b}")

    def __mul__(self, other):
        print(f"Множення: {self} * {other}")
        return Fraction(self.a * other.a, self.b * other.b)

    def __add__(self, other):
        print(f"Додавання: {self} + {other}")
        new_a = self.a * other.b + other.a * self.b
        new_b = self.b * other.b
        return Fraction(new_a, new_b)

    def __sub__(self, other):
        print(f"Віднімання: {self} - {other}")
        new_a = self.a * other.b - other.a * self.b
        new_b = self.b * other.b
        return Fraction(new_a, new_b)

    def __eq__(self, other):
        print(f"Порівняння == : {self} і {other}")
        return self.a == other.a and self.b == other.b

    def __gt__(self, other):
        print(f"Порівняння > : {self} і {other}")
        return self.a * other.b > other.a * self.b

    def __lt__(self, other):
        print(f"Порівняння < : {self} і {other}")
        return self.a * other.b < other.a * self.b

    def __str__(self):
        return f"{self.a}/{self.b}"
```

```python
# створення дробів
f1 = Fraction(2, 3)
f2 = Fraction(3, 6)

print("f1 =", f1)
print("f2 =", f2)

# додавання
f3 = f1 + f2
print("f1 + f2 =", f3)

# віднімання
f4 = f1 - f2
print("f1 - f2 =", f4)

# множення
f5 = f1 * f2
print("f1 * f2 =", f5)

# порівняння
print("f1 == f2:", f1 == f2)
print("f1 > f2:", f1 > f2)
print("f1 < f2:", f1 < f2)

# перевірка скорочення
f6 = Fraction(2, 4)
f7 = Fraction(3, 6)
print("2/4 =", f6)
print("3/6 =", f7)
print("2/4 == 3/6:", f6 == f7)
```
