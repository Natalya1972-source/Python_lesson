ДЗ 5.1. Ім'я змінної

Умова :
1. Рядок може бути ім’ям змінної, якщо одночасно:
2.       не починається з цифри
3.       не містить великих літер
4.       не містить пробілів
5.       не містить знаків пунктуації,
6. крім 
     не є зареєстрованим словом (keyword.kwlist)
     містить не більше одного _

```python
import string
import keyword
```

1 . Варіант 1 - через умовні оператори 

```python
name = input()

if not name:
    print(False)
elif name[0].isdigit():
    print(False)
elif "__" in name:
    print(False)
elif not name.islower():
    print(False)
elif ' ' in name:
    print(False)
elif any(ch in string.punctuation.replace('_', '') for ch in name):
    print(False)
elif name in keyword.kwlist:
    print(False)
else:
    print(True)
```

2. Варіант 2 - через логічні умови 

```python
name = input()

valid = (
    bool(name)
    and not name[0].isdigit()
    and name.islower()
    and ' ' not in name
    and "__" not in name
    and name not in keyword.kwlist
    and not any(ch in string.punctuation.replace('_', '') for ch in name)
)

print(valid)
```
