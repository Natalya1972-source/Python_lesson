правила: 
      прибрати пробіли й пунктуації,
      зробити кожне слово з великої літери,
      додати #, і обрізати до 140 символів.

1. Варіант 1. Реалізовано перетворення введеного рядка у hashtag з видаленням пунктуації, приведенням слів до формату CamelCase та обмеженням довжини до 140 символів

```python
import string

text = input()

for ch in string.punctuation:
    text = text.replace(ch, "")

words = text.split()

hashtag = "#" + "".join(word.capitalize() for word in words)

hashtag = hashtag[:140]

print(hashtag)
```

2. Варіант 2 - Рядок очищено від пунктуації, перетворено у hashtag з форматуванням слів з великої літери та обмеженням довжини до 140 символів.


```python
text = input()

for ch in string.punctuation:
    text = text.replace(ch, "")

hashtag = "#" + "".join(word.capitalize() for word in text.split())
print(hashtag[:140])
```

3. Вапріант 3 - Реалізовано перетворення рядка у hashtag без використання циклу шляхом видалення пунктуації, форматування слів та обмеження довжини результату

```python
text = input()

clean_text = text.translate(str.maketrans("", "", string.punctuation))
hashtag = "#" + "".join(word.capitalize() for word in clean_text.split())

print(hashtag[:140])
```
