1. Варіант 1. Програма створює кортеж літер і словник їхніх позицій, після чого за введеними межами виводить усі символи між двома літерами включно.

```python
import string

letters = tuple(string.ascii_letters)
idx = {c: i for i, c in enumerate(letters)}

a, b = input().split('-')
print(''.join(letters[idx[a]:idx[b] + 1]))
```

2. Варіант (через словник , індекс- літера)


```python
import string

letters = tuple(string.ascii_letters)
idx_to_char = {i: c for i, c in enumerate(letters)}
start_char, end_char = input().split('-')

start = letters.index(start_char)
end = letters.index(end_char)

result = ''
for i in range(start, end + 1):
    result += idx_to_char[i]

print(result)
```
