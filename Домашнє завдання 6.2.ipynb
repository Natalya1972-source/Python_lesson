ДЗ 6.2. Конвертер із числа в дату

1 Варіант.Через функцію divmod 

```python
seconds = int(input())

day_sec = 24 * 60 * 60
hour_sec = 60 * 60

days, remainder = divmod(seconds, day_sec)
hours, remainder = divmod(remainder, hour_sec)
minutes, secs = divmod(remainder, 60)

if days % 10 == 1 and days % 100 != 11:
    day_word = "день"
elif 2 <= days % 10 <= 4 and not (12 <= days % 100 <= 14):
    day_word = "дні"
else:
    day_word = "днів"

print(f"{days} {day_word}, {str(hours).zfill(2)}:{str(minutes).zfill(2)}:{str(secs).zfill(2)}")
```

2. Варіант. Тільки базові операції // і %

```python
seconds = int(input())

days = seconds // 86400
seconds = seconds % 86400

hours = seconds // 3600
seconds = seconds % 3600

minutes = seconds // 60
secs = seconds % 60

if days % 10 == 1 and days % 100 != 11:
    day_word = "день"
elif 2 <= days % 10 <= 4 and not (12 <= days % 100 <= 14):
    day_word = "дні"
else:
    day_word = "днів"

print(f"{days} {day_word}, {str(hours).zfill(2)}:{str(minutes).zfill(2)}:{str(secs).zfill(2)}")
```
