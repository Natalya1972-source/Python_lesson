ДЗ 6.3. Добуток чисел

Користувач вводить ціле число
Поки число більше 9:
розбиваємо його на цифри
перемножуємо всі цифри
замінюємо число на отриманий добуток
Коли число стало ≤ 9 — виводимо результат

Варіант 1. Програма послідовно перемножує цифри введеного числа, доки результат не стане однозначним.

```python
n = int(input())

while n > 9:
    product = 1
    for digit in str(n):
        product *= int(digit)
    n = product

print(n)
```

Варіант 2. Без перетворення числа в рядок, тільки з математикою (// і %)

```python
n = int(input())

while n > 9:
    product = 1
    temp = n

    while temp > 0:
        digit = temp % 10
        product *= digit
        temp //= 10

    n = product

print(n)
```

Варіант 3 (кортеж + словник)

```python
n = int(input())

while n > 9:
    digits = tuple(int(d) for d in str(n))
    prod = 1
    for d in digits:
        prod *= d
    n = prod

print(n)
```
