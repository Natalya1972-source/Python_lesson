Варіант 1. реалізовано функцію say_hi, яка приймає ім’я та вік, повертає рядок привітання відповідно до умов і успішно проходить перевірку тестами.

```python
def say_hi(name, age):
    return f"Hi. My name is {name} and I'm {age} years old"


assert say_hi("Alex", 32) == "Hi. My name is Alex and I'm 32 years old", 'Test1'
assert say_hi("Frank", 68) == "Hi. My name is Frank and I'm 68 years old", 'Test2'

print('ОК')
```

Варіант 2 . Виконала в межах умов, з використанням функцій та множин (set), без зміни очікуваного результату.

```python
def say_hi(name, age):
    name = "".join(set([name]))  
    return f"Hi. My name is {name} and I'm {age} years old"


assert say_hi("Alex", 32) == "Hi. My name is Alex and I'm 32 years old", 'Test1'
assert say_hi("Frank", 68) == "Hi. My name is Frank and I'm 68 years old", 'Test2'

print("ОК")
```

Варіант 3. (використала format())

```python
def say_hi(name, age):
    result = "Hi. My name is {} and I'm {} years old"
    return result.format(name, age)


assert say_hi("Alex", 32) == "Hi. My name is Alex and I'm 32 years old", 'Test1'
assert say_hi("Frank", 68) == "Hi. My name is Frank and I'm 68 years old", 'Test2'

print("ОК")
```

Варіант 4. Через конкатенацію рядків 

```python
def say_hi(name, age):
    return "Hi. My name is " + name + " and I'm " + str(age) + " years old"


assert say_hi("Alex", 32) == "Hi. My name is Alex and I'm 32 years old", 'Test1'
assert say_hi("Frank", 68) == "Hi. My name is Frank and I'm 68 years old", 'Test2'

print("ОК")
```
