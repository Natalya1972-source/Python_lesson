ДЗ 7.4. Пошук спільних елементів

Варіант 1. Функція формує множини чисел, кратних 3 і 5 у діапазоні від 0 до 99, та повертає їх перетин як результат.

```python
def common_elements():
    multiples_3 = {x for x in range(100) if x % 3 == 0}
    multiples_5 = {x for x in range(100) if x % 5 == 0}
    return multiples_3 & multiples_5


assert common_elements() == {0, 75, 45, 15, 90, 60, 30}
print('ОК')
```

Варіант 2. Створюємо два списки через list comprehension, перетворюємо їх у множини, знаходимо перетин множин через intersection(), повертаємо результат

```python
def common_elements():
    list_3 = [x for x in range(100) if x % 3 == 0]
    list_5 = [x for x in range(100) if x % 5 == 0]

    set_3 = set(list_3)
    set_5 = set(list_5)

    return set_3.intersection(set_5)


assert common_elements() == {0, 75, 45, 15, 90, 60, 30}
print('ОК')
```

Варіант 3 . Два генератори одразу створюють множини, оператор & знаходить їх перетин, результат — множина спільних елементів

```python
def common_elements():
    return set(x for x in range(100) if x % 3 == 0) & set(x for x in range(100) if x % 5 == 0)


assert common_elements() == {0, 75, 45, 15, 90, 60, 30}
print('ОК')
```

Варіант 4. Створюємо порожні списки, циклом перебираємо числа від 0 до 99, додаємо числа, кратні 3 і 5, перетворюємо списки на множини, знаходимо перетин

```python
def common_elements():
    list_3 = []
    list_5 = []

    for i in range(100):
        if i % 3 == 0:
            list_3.append(i)
        if i % 5 == 0:
            list_5.append(i)

    set_3 = set(list_3)
    set_5 = set(list_5)

    return set_3.intersection(set_5)


assert common_elements() == {0, 75, 45, 15, 90, 60, 30}
print('ОК')
```
