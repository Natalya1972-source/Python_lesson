ДЗ 8.1. Додати 1 до числа


1. Варіант 1. (int) список цифр перетворюється на число, до нього додається одиниця, після чого результат знову розбивається на список цифр.

```python
def add_one(some_list):
    number = int(''.join(map(str, some_list)))
    number += 1
    return [int(digit) for digit in str(number)]

assert add_one([1, 2, 3, 4]) == [1, 2, 3, 5], 'Test1'
assert add_one([9, 9, 9]) == [1, 0, 0, 0], 'Test2'
assert add_one([0]) == [1], 'Test3'
assert add_one([9]) == [1, 0], 'Test4'
print("ОК")
```

2. Варіант (без int) додавання одиниці виконується вручну з кінця списку з урахуванням переносу розрядів, як у звичайному додаванні чисел.

```python
def add_one(some_list):
    result = some_list[:]
    i = len(result) - 1

    while i >= 0:
        if result[i] < 9:
            result[i] += 1
            return result
        else:
            result[i] = 0
            i -= 1

    return [1] + result


assert add_one([1, 2, 3, 4]) == [1, 2, 3, 5], 'Test1'
assert add_one([9, 9, 9]) == [1, 0, 0, 0], 'Test2'
assert add_one([0]) == [1], 'Test3'
assert add_one([9]) == [1, 0], 'Test4'
print("ОК")
```

3. Варіант (join + int)

```python
def add_one(some_list):
    s = ''.join(str(x) for x in some_list)
    return list(map(int, str(int(s) + 1)))


assert add_one([1, 2, 3, 4]) == [1, 2, 3, 5], 'Test1'
assert add_one([9, 9, 9]) == [1, 0, 0, 0], 'Test2'
assert add_one([0]) == [1], 'Test3'
assert add_one([9]) == [1, 0], 'Test4'
print("ОК")
```

4. Варіант (через роботу з рядком)

```python
def add_one(some_list):
    s = ''.join(map(str, some_list))
    s = str(int(s) + 1)
    return [int(x) for x in s]


assert add_one([1, 2, 3, 4]) == [1, 2, 3, 5], 'Test1'
assert add_one([9, 9, 9]) == [1, 0, 0, 0], 'Test2'
assert add_one([0]) == [1], 'Test3'
assert add_one([9]) == [1, 0], 'Test4'
print("ОК")
```
