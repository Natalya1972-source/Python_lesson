ДЗ 8.3. Унікальне число

1. Варіант (через count)

```python
def find_unique_value(some_list):
    for value in some_list:
        if some_list.count(value) == 1:
            return value


assert find_unique_value([1, 2, 1, 1]) == 2, 'Test1'
assert find_unique_value([2, 3, 3, 3, 5, 5]) == 2, 'Test2'
assert find_unique_value([5, 5, 5, 2, 2, 0.5]) == 0.5, 'Test3'
print("ОК")
```

2. Варіант (через set)

```python
def find_unique_value(some_list):
    for value in set(some_list):
        if some_list.count(value) == 1:
            return value


assert find_unique_value([1, 2, 1, 1]) == 2, 'Test1'
assert find_unique_value([2, 3, 3, 3, 5, 5]) == 2, 'Test2'
assert find_unique_value([5, 5, 5, 2, 2, 0.5]) == 0.5, 'Test3'
print("ОК")
```

3. Варіант (через словник)

```python
def find_unique_value(some_list):
    counts = {}
    for value in some_list:
        counts[value] = counts.get(value, 0) + 1

    for key, count in counts.items():
        if count == 1:
            return key


assert find_unique_value([1, 2, 1, 1]) == 2, 'Test1'
assert find_unique_value([2, 3, 3, 3, 5, 5]) == 2, 'Test2'
assert find_unique_value([5, 5, 5, 2, 2, 0.5]) == 0.5, 'Test3'
print("ОК")
```

4. Варіант через Counter

```python
from collections import Counter

def find_unique_value(some_list):
    counter = Counter(some_list)
    for key, value in counter.items():
        if value == 1:
            return key


assert find_unique_value([1, 2, 1, 1]) == 2, 'Test1'
assert find_unique_value([2, 3, 3, 3, 5, 5]) == 2, 'Test2'
assert find_unique_value([5, 5, 5, 2, 2, 0.5]) == 0.5, 'Test3'
print("ОК")
```

5. Варіант через filter

```python
def find_unique_value(some_list):
    return list(filter(lambda x: some_list.count(x) == 1, some_list))[0]


assert find_unique_value([1, 2, 1, 1]) == 2, 'Test1'
assert find_unique_value([2, 3, 3, 3, 5, 5]) == 2, 'Test2'
assert find_unique_value([5, 5, 5, 2, 2, 0.5]) == 0.5, 'Test3'
print("ОК")
```
