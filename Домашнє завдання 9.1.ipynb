ДЗ 9.1. Визначити популярність певних слів у тексті


Варіант 1. Текст перевела в нижній регістр і розбила на слова.
Далі через словникове включення порахувала кількість входжень кожного слова методом count(), який повертає 0, якщо слово відсутнє.

```python
def popular_words(text, words):
    text = text.lower().split()
    return {word: text.count(word) for word in words}


assert popular_words(
    '''When I was One I had just begun
       When I was Two I was nearly new''',
    ['i', 'was', 'three', 'near']
) == {
    'i': 4,
    'was': 3,
    'three': 0,
    'near': 0
}, 'Test1'

print('OK')
```

Варіант 2. Текст перевела в нижній регістр і розбила на слова.
За допомогою Counter порахувала кількість усіх слів, а потім сформувала словник лише для потрібних слів.
Якщо слова немає — Counter повертає 0.

```python
from collections import Counter

def popular_words(text, words):
    text_words = text.lower().split()
    counter = Counter(text_words)
    return {word: counter[word] for word in words}


assert popular_words(
    '''When I was One I had just begun
       When I was Two I was nearly new''',
    ['i', 'was', 'three', 'near']
) == {
    'i': 4,
    'was': 3,
    'three': 0,
    'near': 0
}, 'Test1'

print('OK')
```

Варіант 3. Створила словник зі словами та нульовими значеннями.
Пройшлася по всіх словах тексту й збільшувала лічильник, якщо слово є у списку шуканих.

```python
def popular_words(text, words):
    result = {word: 0 for word in words}
    for w in text.lower().split():
        if w in result:
            result[w] += 1
    return result


assert popular_words(
    '''When I was One I had just begun
       When I was Two I was nearly new''',
    ['i', 'was', 'three', 'near']
) == {
    'i': 4,
    'was': 3,
    'three': 0,
    'near': 0
}, 'Test1'

print('OK')
```
