ДЗ 9.2. Різниця між числами

Варіант 1. Функція приймає довільну кількість аргументів через *args.
Якщо аргументів немає — повертає 0.В іншому випадку обчислює різницю між максимальним і мінімальним значенням, використовуючи round для коректної роботи з float.

```python
def difference(*args):
    if not args:
        return 0
    return round(max(args) - min(args), 2)


assert difference(1, 2, 3) == 2, 'Test1'
assert difference(5, -5) == 10, 'Test2'
assert difference(10.2, -2.2, 0, 1.1, 0.5) == 12.4, 'Test3'
assert difference() == 0, 'Test4'

print('OK')
```

Варіант 2. Якщо аргументів немає, функція повертає 0.
Початкові мінімум і максимум ініціалізую першим елементом, далі проходжу по всіх аргументах циклом і оновлюю значення.
Наприкінці повертаю різницю з округленням.

```python
def difference(*args):
    if not args:
        return 0

    min_val = args[0]
    max_val = args[0]

    for x in args:
        if x < min_val:
            min_val = x
        if x > max_val:
            max_val = x

    return round(max_val - min_val, 2)


assert difference(1, 2, 3) == 2, 'Test1'
assert difference(5, -5) == 10, 'Test2'
assert difference(10.2, -2.2, 0, 1.1, 0.5) == 12.4, 'Test3'
assert difference() == 0, 'Test4'

print('OK')
```

Варіант 3. Використала reduce, щоб за один прохід по аргументах знайти мінімальне і максимальне значення.

```python
from functools import reduce

def difference(*args):
    if not args:
        return 0

    min_val, max_val = reduce(
        lambda acc, x: (min(acc[0], x), max(acc[1], x)),
        args[1:],
        (args[0], args[0])
    )

    return round(max_val - min_val, 2)


assert difference(1, 2, 3) == 2, 'Test1'
assert difference(5, -5) == 10, 'Test2'
assert difference(10.2, -2.2, 0, 1.1, 0.5) == 12.4, 'Test3'
assert difference() == 0, 'Test4'

print('OK')
```

Варіант 4. Функція приймає довільну кількість аргументів.
Якщо аргументів немає — повертає 0, інакше обчислює різницю між максимальним і мінімальним значенням з округленням.

```python
def difference(*args):
    return 0 if not args else round(max(args) - min(args), 2)


assert difference(1, 2, 3) == 2, 'Test1'
assert difference(5, -5) == 10, 'Test2'
assert difference(10.2, -2.2, 0, 1.1, 0.5) == 12.4, 'Test3'
assert difference() == 0, 'Test4'

print('OK')
```

Додатковий варіант. Це дічь, але мені було дуже цікаво чи вийде. Не оцінюйте , просто гралась. Це рекурсивний , як демонстрація підходу.
Рекурсія використовується для послідовного знаходження мінімального і максимального значення, після чого повертається їх різниця.
Для практичного використання доцільніше застосовувати цикл або max/min.

```python
difference = lambda *a: 0 if not a else round(
    (lambda f: f(f, a, a[0], a[0]))(
        lambda self, nums, mn, mx:
            (mn, mx) if not nums
            else self(
                self,
                nums[1:],
                mn if mn < nums[0] else nums[0],
                mx if mx > nums[0] else nums[0]
            )
    )[1]
    -
    (lambda f: f(f, a, a[0], a[0]))(
        lambda self, nums, mn, mx:
            (mn, mx) if not nums
            else self(
                self,
                nums[1:],
                mn if mn < nums[0] else nums[0],
                mx if mx > nums[0] else nums[0]
            )
    )[0],
    2
)


assert difference(1, 2, 3) == 2, 'Test1'
assert difference(5, -5) == 10, 'Test2'
assert difference(10.2, -2.2, 0, 1.1, 0.5) == 12.4, 'Test3'
assert difference() == 0, 'Test4'

print('OK')
```
